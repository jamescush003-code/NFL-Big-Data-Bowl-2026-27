```python
import pandas as pd

df = pd.read_csv('combine_tracking.csv',
                 usecols=['entity_type', 'drill_type', 'drill_name',
                          'event_id', 'nfl_id', 'draft_year'])

summary = (df[df.entity_type == 'PLAYER']
           .groupby(['drill_type', 'drill_name'])
           .agg(attempts=('event_id', 'nunique'),
                players=('nfl_id', 'nunique'),
                years=('draft_year', lambda x: sorted(x.unique())))
           .sort_values(['drill_type', 'attempts'], ascending=[True, False]))

pd.set_option('display.max_rows', None)
summary
```
